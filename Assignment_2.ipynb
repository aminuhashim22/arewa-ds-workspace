```python
#Area Calculator

Length = float(input("Enter the length of the rectangle: "))
Width = float(input("Enter the width of the rectangle: "))
Area = Length * Width
Parameter = 2 * Length + 2 * Width
print("The area of the rectangle is:", Area)
print("The perimeter of the rectangle is:", Parameter)
```

```python
#Shopping Bill Calculator

first_price = float(input("Enter the price of the first item: "))
second_price = float(input("Enter the price of the second item: "))
third_price = float(input("Enter the price of the third item: "))


subtotal = first_price + second_price + third_price
sales_tax = subtotal * 0.06
final_total = subtotal + sales_tax
print("The subtotal is:", subtotal)
print("The sales tax is :", sales_tax)
print("The final total is :", final_total)
```
