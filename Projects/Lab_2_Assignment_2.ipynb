```python
### Exercise 3. Change maker

#Given an amount in kobo, work out how many 100 naira notes, 50 naira notes, and leftover kobo it contains.
T#here are 100 kobo in 1 naira.
```

```python
amountKobo = 27350
amountNaira = amountKobo // 100
notes100 = amountNaira // 100
remaininginNaira = amountNaira % 100
notes50  = remaininginNaira // 50
leftoverKobo = amountKobo % 100

print(notes100, "x 100 naira,", notes50, "x 50 naira, and", leftoverKobo, "kobo left over")
```

```python
### Exercise 5. Compound interest

#Compute the balance after `years` years of compound interest, using the formula

#balance = principal * (1 + rate) ** years

#Round the answer to two decimal places.
```

```python
principal = 50000.0
rate = 0.12
years = 5

balance = round(principal * (1 + rate) ** years, 2)

print("Balance after", years, "years:", balance)
```

```python
### Exercise 6. Challenge, no automatic check

#Write a complete, documented program in the cell below that:

#1. asks the user for the radius of a circle
#2. computes the circumference and the area
#3. prints both, each labelled, using `str` conversion
#4. has a docstring and at least two comments

```

```python
import math
radius = float(input("Enter the radius of the circle: "))
circumference = 2 * math.pi * radius 
area = math.pi * radius ** 2

print ("Circumference:" + str(round(circumference, 2)))
print("Area:" + str(round(area, 2)))
```
